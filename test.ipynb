```python
import os
from PIL import Image
image_root = 'data/messy_rooms/large_corridor_50/images_copy'
save_root = 'data/messy_rooms/large_corridor_50/images'
image_names = os.listdir(image_root)
image_names.sort()
os.makedirs(save_root, exist_ok=True)

for image_name in image_names:

    Image.open(os.path.join(image_root, image_name)).resize((510, 510)).save(os.path.join(save_root, image_name))
```
